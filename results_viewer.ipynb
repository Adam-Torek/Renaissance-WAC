```python
def show_metric_results_table(table_path, metric_value_to_show):
    table_to_show = pd.read_csv(table_path)
    reformatted_table = table_to_show[table_to_show["metric_name"] == metric_value_to_show]["metric_value"].apply(lambda x: round(x * 100, 3))
    table_to_show.loc[table_to_show["metric_name"] == metric_value_to_show, "metric_value"] = reformatted_table
    print(table_to_show[table_to_show["metric_name"] == metric_value_to_show])
```

```python
def plot_test_results(test_df):
    phases = list(test_df["phase"].unique())
    metric_names = list(test_df["metric_name"].unique())
    for phase in phases:
        for metric_name in metric_names:
            df_to_plot = test_df.loc[(test_df["phase"] == phase) & (test_df["metric_name"] == metric_name)]
            df_to_plot.plot.line(x="epoch", y="metric_value")
            plt.xlabel("Epoch")
            plt.ylabel(metric_name)
            plt.title(f"Metric Result {metric_name} for {phase} Split")
            plt.show()
```

```python
wac_embedding_results = pd.read_csv("result/eval_ref_twotower_electrasmall_wac_text_pretrained_embeddings_ref_seed0_is224_ps16_bs8_pgbs8_ts100000/version_6/eval.csv")
plot_test_results(wac_embedding_results)
```

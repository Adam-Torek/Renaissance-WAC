```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
itm_results = pd.read_csv("result/wac_itm_twotower_electrasmall_deit_small_seed0_is224_ps16_bs256_pgbs256_ts100000/version_0/eval.csv")
ref_results = pd.read_csv("result/wac_ref_twotower_electrasmall_deit_small_seed0_is224_ps16_bs256_pgbs256_ts100000/version_0/eval.csv")
ref_bbox_results = pd.read_csv("result/wac_ref_bbox_twotower_electrasmall_deit_small_seed0_is224_ps16_bs256_pgbs256_ts100000/version_0/eval.csv")
```

```python
def plot_test_results(test_df):
    phases = list(test_df["phase"].unique())
    metric_names = list(test_df["metric_name"].unique())
    for phase in phases:
        for metric_name in metric_names:
            df_to_plot = test_df.loc[(test_df["phase"] == phase) & (test_df["metric_name"] == metric_name)]
            df_to_plot.plot.line(x="epoch", y="metric_value")
            plt.xlabel("Epoch")
            plt.ylabel(metric_name)
            plt.title(f"Metric Result {metric_name} for {phase} Split")
            plt.show()
```

```python
plot_test_results(itm_results)
```

```python
plot_test_results(ref_results)
```

```python
plot_test_results(ref_bbox_results)
```
